```julia
using LinearAlgebra

function P(m, i, j)
    A = diagm(0 => ones(m))
    A[i, :], A[j, :] = A[j, :], A[i, :]
    return A
end

function Q(m, i, c)
    A = diagm(0 => ones(m))
    A[i, i] = c
    return A
end

function R(m, i, j, c)
    A = diagm(0 => ones(m))
    A[i, j] = c
    return A
end
```

```julia
function CR(A)
    # A.rref() から行列部分を取得
    E = A.rref()[1]
    
    # ランクの取得
    r = Int(rank(A))
    
    # ピボット列のインデックスを取得（ここまでは型を気にせず進める）
    col = [findfirst(!iszero, E[i, :]) for i in 1:r]
    
    # --- ここで型を強制的に Matrix{Sym} に変換する ---
    # Array{Sym}(...) で Julia 側の Sym 型行列として再構成します
    C = Array{Sym}(A[:, col])
    R = Array{Sym}(E[1:r, :])
    
    return C, R
end
```

```julia
C, R₁ = CR(A)
```

```julia
C
```

```julia
R₁
```

```julia
A == C * R₁
```

$A,B$ は $3$ 次正方行列の中から探すことにする。  
$\operatorname{rank}(A)=3$ or $\operatorname{rank}(B)=3$ のときには、定理3.3.8より、 $\operatorname{rank}(AB)=\operatorname{rank}(BA)=3$ になってしまう。  
$\operatorname{rank}(A)\leq 1$ or $\operatorname{rank}(B)\leq 1$ のときには、定理3.3.10より、 $\operatorname{rank}(AB)\leq 1$ and $\operatorname{rank}(BA)\leq 1$ となってしまい、よりつまらない例しか探せない。   
そこで、 $\operatorname{rank}(A)=\operatorname{rank}(B)=2$ となるような行列で、 $\operatorname{rank}(AB)=1$ and $\operatorname{rank}(BA)=2$ となるようなものを探すことにする。
