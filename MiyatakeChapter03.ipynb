```julia
using LinearAlgebra

function P(m, i, j)
    A = diagm(0 => ones(m))
    A[i, :], A[j, :] = A[j, :], A[i, :]
    return A
end

function Q(m, i, c)
    A = diagm(0 => ones(m))
    A[i, i] = c
    return A
end

function R(m, i, j, c)
    A = diagm(0 => ones(m))
    A[i, j] = c
    return A
end
```

```julia
function CR(A)
    # A.rref() から行列部分を取得
    E = A.rref()[1]
    
    # ランクの取得
    r = Int(rank(A))
    
    # ピボット列のインデックスを取得（ここまでは型を気にせず進める）
    col = [findfirst(!iszero, E[i, :]) for i in 1:r]
    
    # --- ここで型を強制的に Matrix{Sym} に変換する ---
    # Array{Sym}(...) で Julia 側の Sym 型行列として再構成します
    C = Array{Sym}(A[:, col])
    R = Array{Sym}(E[1:r, :])
    
    return C, R
end
```

```julia
C, R₁ = CR(A)
```

```julia
C
```

```julia
R₁
```

```julia
A == C * R₁
```

## 3.6

$A$ を $m\times n$ 行列とする。  
$B$ を $p\times q$ 行列とする。  
$A=\begin{bmatrix}
   a_1 & a_2 & \cdots & a_n
\end{bmatrix},
B=\begin{bmatrix}
   b_1 & b_2 & \cdots & b_q
\end{bmatrix}$ と書く。  
$A\otimes B =
\begin{bmatrix}
a_1\otimes b_1 & a_1\otimes b_2 & \cdots & a_1\otimes b_q & a_2\otimes b_1 & a_2\otimes b_2 & \cdots & a_2\otimes b_q & \cdots & a_n\otimes b_1 & a_n\otimes b_2 \cdots & a_n\otimes b_q 
\end{bmatrix}$ である。  
$\operatorname{rank}(A)=r, \operatorname{rank}(B)=s$ とする。  
$a_{i_1},\dots,a_{i_r}$, $b_{j_1},\dots,b_{j_s}$ をそれぞれ一次独立な列ベクトルの列とする。  
$a_{i_1}\otimes b_{j_1}, a_{i_1}\otimes b_{j_2}, \cdots, a_{i_1}\otimes b_{j_s}, a_{i_2}\otimes b_{j_1}, a_{i_2}\otimes b_{j_2}, \cdots, a_{i_2}\otimes b_{j_s}, \cdots, a_{i_r}\otimes b_{j_1}, a_{i_r}\otimes b_{j_2}, \cdots, a_{i_r}\otimes b_{j_s}$ が一次独立であることを以下で示す。  
$c_{11} a_{i_1}\otimes b_{j_1} + c_{12} a_{i_1}\otimes b_{j_2} + \cdots + c_{1s} a_{i_1}\otimes b_{j_s} + c_{21} a_{i_2}\otimes b_{j_1} + c_{22} a_{i_2}\otimes b_{j_2} + \cdots + c_{2s} a_{i_2}\otimes b_{j_s} + \cdots + c_{r1} a_{i_r}\otimes b_{j_1} + c_{r2} a_{i_r}\otimes b_{j_2} + \cdots + c_{rs} a_{i_r}\otimes b_{j_s} = 0$ とする。  
p64 定理2.4.5を使って整理する。  
$a_{i_1}\otimes (c_{11} b_{j_1} + c_{12} b_{j_2} + \cdots + c_{1s} b_{j_s}) +
a_{i_2}\otimes (c_{21} b_{j_1} + c_{22} b_{j_2} + \cdots + c_{2s} b_{j_s}) + 
\cdots
a_{i_r}\otimes (c_{r1} b_{j_1} + c_{r2} b_{j_2} + \cdots + c_{rs} b_{j_s}) = 0$ である。  
$a_{i_1}, \dots, a_{i_r}$ が一次独立であるとき、 $\sum_k a_{i_k}\otimes d_{i_k} = 0$ ならば $d_{i_k} = 0$ が成り立つことは $\sum_k a_{i_k}\otimes d_{i_k}$ を具体的に書いてみれば容易にわかる。  
よって、 $c_{k1} b_{j_1} + c_{k2} b_{j_2} + \cdots + c_{ks} b_{j_s}=0$ である。  
$b_{j_1},\dots,b_{j_s}$ は一次独立な列なので、 $c_{ij}=0$ である。  
よって、 $a_{i_1}\otimes b_{j_1}, a_{i_1}\otimes b_{j_2}, \cdots, a_{i_1}\otimes b_{j_s}, a_{i_2}\otimes b_{j_1}, a_{i_2}\otimes b_{j_2}, \cdots, a_{i_2}\otimes b_{j_s}, \cdots, a_{i_1}\otimes b_{j_s}, a_{i_2}\otimes b_{j_s}, \cdots, a_{i_r}\otimes b_{j_s}$ は一次独立である。   
$a_i$ を $A$ の任意の列ベクトルとする。  
$b_j$ を $B$ の任意の列ベクトルとする。  
$a_i = \alpha_1 a_{i_1} + \cdots + \alpha_r a_{i_r}$ と書ける。  
$b_j = \beta_1 a_{j_1} + \cdots + \beta_s a_{j_s}$ と書ける。  
\begin{align*}
a_i\otimes b_j&=(\alpha_1 a_{i_1} + \cdots + \alpha_r a_{i_r})\otimes(\beta_1 a_{j_1} + \cdots + \beta_s a_{j_s})\\
&=\alpha_1\beta_1 (a_{i_1}\otimes b_{j_1}) + \alpha_1\beta_1 (a_{i_1}\otimes b_{j_2}) + \cdots + \alpha_1\beta_s (a_{i_1}\otimes b_{j_s}) + \alpha_2\beta_1 (a_{i_2}\otimes b_{j_1}) + \alpha_2\beta_2 (a_{i_2}\otimes b_{j_2}) + \cdots + \alpha_2\beta_s (a_{i_2}\otimes b_{j_s}) + \cdots + \alpha_r\beta_1 (a_{i_r}\otimes b_{j_1}) + \alpha_r\beta_2 (a_{i_r}\otimes b_{j_2}) + \cdots + \alpha_r\beta_s (a_{i_r}\otimes b_{j_s}) = 0
\end{align*}
である。  
よって、 $\operatorname{rank}(A\otimes B)=rs$ である。
