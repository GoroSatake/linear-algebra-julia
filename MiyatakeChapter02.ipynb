$A,B$が正則であるとする。  
$(A\otimes B) (A^{-1}\otimes B^{-1})=(AA^{-1})\otimes (BB^{-1})=I\otimes I=I$である。  
$(A^{-1}\otimes B^{-1}) (A\otimes B)=(A^{-1}A)\otimes (B^{-1}B)=I\otimes I=I$である。

$y\neq 0$とする。  
$(A\otimes B)(x\otimes y)=(Ax)\otimes (By)=0\otimes (By)=0$である。  
$x\otimes y\neq 0$であるから、$A\otimes B$は正則ではない。これは矛盾である。  
よって、$A$は正則である。

## 2.5

上三角行列のサイズに関する数学的帰納法によって証明する。  
$n=1$ のときには明らかに成り立つ。  
$n=k$ のときに成り立つと仮定し、$n=k+1$ の場合を考える。  
$A_{k+1}$ をサイズが$k+1$の正則な上三角行列とする。  
$A_{k+1}$の対角成分がすべて$0$でないことは容易に分かる。  
$A_k$ をサイズが $k$ の上三角行列として、
$$
B_{k+1}A_{k+1}=
\left[
\begin{array}{c:c} 
B_k & {\begin{array}{c} * \\ * \\ \vdots \\ * \end{array}} \\ 
\hdashline
{\begin{array}{cccc} b_{k+1,1} & b_{k+1,2} & \cdots & b_{k+1,k} \end{array}} & b_{k+1,k+1} 
\end{array}
\right]
\left[
\begin{array}{c:c} 
A_k & {\begin{array}{c} * \\ * \\ \vdots \\ * \end{array}} \\ 
\hdashline
{\begin{array}{cccc} 0 & 0 & \cdots & 0 \end{array}} & a_{k+1,k+1} 
\end{array}
\right]
=
\left[
\begin{array}{c:c} 
I_k & {\begin{array}{c} 0 \\ 0 \\ \vdots \\ 0 \end{array}} \\ 
\hdashline
{\begin{array}{cccc} 0 & 0 & \cdots & 0 \end{array}} & 1 
\end{array}
\right]
=I_{k+1}
$$
と書ける。  
$B_kA_k=I_k$ であるから、$B_k=A_k^{-1}$は帰納法の仮定により上三角行列である。  
$$
\left[
\begin{array}{cccc} b_{k+1,1} & b_{k+1,2} & \cdots & b_{k+1,k} \end{array}
\right]
A_k
=
\left[
\begin{array}{cccc} 0 & 0 & \cdots & 0 \end{array}
\right]
$$
である。  
$A_k$の対角成分はすべて$0$でないから、
$$
\left[
\begin{array}{cccc} b_{k+1,1} & b_{k+1,2} & \cdots & b_{k+1,k} \end{array}
\right]=
\left[
\begin{array}{cccc} 0 & 0 & \cdots & 0 \end{array}
\right]
$$ であることが容易に分かる。  
以上から、$B_{k+1}$は上三角行列である。

```julia
n = 5
A = diagm(0 => -ones(n-1), 2 => ones(n-3))
e₁ = zeros(n-1)
e₁[n-2] = 1
e₂ = zeros(n-1)
e₂[n-1] = 1
[A e₁ e₂]
```

```julia
function derivative_approx(n)
    Δx = 1/n
    x = 0:Δx:1
    f = sin.(2*π*x)
    A = diagm(0 => -ones(n-1), 2 => ones(n-3))
    e₁ = zeros(n-1)
    e₁[n-2] = 1
    e₂ = zeros(n-1)
    e₂[n-1] = 1
    A = 1/(2*Δx) * [A e₁ e₂]
    g = A * f
    return 2, n, x, g
end
```

```julia
function second_derivative_approx(n)
    (i, j, x, g) = derivative_approx(n)
    A = diagm(0 => -ones(n-3), 2 => ones(n-5))
    e₁ = zeros(n-3)
    e₁[n-4] = 1
    e₂ = zeros(n-3)
    e₂[n-3] = 1
    Δx = 1/n
    B = 1/(2*Δx) * [A e₁ e₂]
    h = B * g
    return 3, n-1, x, h
end
```

```julia
function third_derivative_approx(n)
    (i, j, x, g) = second_derivative_approx(n)
    A = diagm(0 => -ones(n-5), 2 => ones(n-7))
    e₁ = zeros(n-5)
    e₁[n-6] = 1
    e₂ = zeros(n-5)
    e₂[n-5] = 1
    Δx = 1/n
    B = 1/(2*Δx) * [A e₁ e₂]
    h = B * g
    return 4, n-2, x, h
end
```

```julia
using Plots
using LinearAlgebra
using LaTeXStrings

x_ref = 0:1/1000:1
g_ref = -4*π^2*sin.(2*π*x_ref)
plot(x_ref, g_ref, color =:black, xlabel = "x", label = L"f^{''} (x)=-4\pi^2\sin (2\pi x)")
i, j, x, h = second_derivative_approx(10)
plot!((x[i:j], h), seriestype =:scatter, markershape =:rect, markercolor =:white, label = L"\Delta x=0.1")
i, j, x, h = second_derivative_approx(20)
plot!((x[i:j], h), seriestype =:scatter, markershape =:circ, markercolor =:gray, label = L"\Delta x=0.05")
```
