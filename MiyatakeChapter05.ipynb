# 演習問題

## 5.5

$\alpha,\beta\in\mathbb{R}-\{0\}$ とする。
$(\alpha x)\oplus (\beta y)=x\oplus y$ が成り立つことに注意する。  
$\alpha>0$ とする。 $a\odot (\alpha x)=a\odot x$ が成り立つことに注意する。  
この2つの性質を以下、断りなく使うことにする。

(1)  
\begin{align*}
(x\oplus y)\oplus z
&=
\left[x_1y_1, \dots, x_ny_n\right]
\oplus
\left[z_1, \dots, z_n\right]
=
\left[(x_1y_1)z_1, \dots, (x_ny_n)z_n\right]\\
&=
\left[x_1(y_1z_1), \dots, x_n(y_nz_n)\right]
=
\left[x_1, \dots, x_n\right]
\oplus
\left[y_1z_1, \dots, y_nz_n\right]
=
x\oplus (y\oplus z)
\end{align*}

(2) 明らか。

(3)  
$0 = \left[\frac{1}{n}, \dots, \frac{1}{n}\right]$ である。  
実際、  
$0\oplus x = [1, \dots, 1]\oplus [x_1,\dots,x_n]=x$ である。

(4)  
$x' = \left[\frac{x_1^{-1}}{\sum x_i^{-1}}, \dots, \frac{x_n^{-1}}{\sum x_i^{-1}}\right]$ である。  
実際、  
$x\oplus x' = [x_1,\dots,x_n]\oplus [x_1^{-1},\dots,x_n^{-1}]=[\frac{1}{n},\dots,\frac{1}{n}]=0$ である。

(5)  
$(a\odot x)\oplus (b\odot x) = [x_1^a,\dots,x_n^a]\oplus [x_1^b,\dots,x_n^b]=[Ax_1^ax_1^b,\dots,Ax_n^ax_n^b]=[Ax_1^{a+b},\dots,Ax_1^{a+b}]
=(a+b)\odot x$ である。  
ただし、 $A=\frac{1}{\sum x_i^{a+b}}$ である。


(6)  
$(a\odot x)\oplus (a\odot y) = [x_1^a,\dots,x_n^a]\oplus [y_1^a,\dots,y_n^a]=[Ax_1^ay_1^a,\dots,Ax_n^ay_n^a]=a\odot (x\oplus y)$ である。  
ただし、$A=\frac{1}{x_i^ay_i^a}$ である。

(7)  
$a\odot(b\odot x) = a\odot[x_1^b,\dots,x_n^b] = \left[A(x_1^b)^a, \dots, A(x_n^b)^a\right]=(ab)\odot x$ である。  
ただし、$A=\frac{1}{\sum (x_i^b)^a}$ である。

(8)  
明らか。
